# Choosing parameters for discretization 

We don't want to have "ladder-like" trajectories on subsampling/discretization as that will introduce artefacts in swimming directions (i.e. biased towards $\pm\pi$, $\pm\pi/2$)

As older fish swim faster, we need to drop fewer frames to achieve this.

If it doesn't slow down predictor computation by a lot, it would be best to have 1 grid_unit = 1 mm

*Note: having 1 grid unit = 1mm seems like an overkill for older fish (in the larger arena). Reducing this*

# Parameters for predictors 
The parameters are decided based on analysis of raw data (offline on PP's computer). Using the parameters directly here. 

Note that care must be taken to ensure appropriate unit conversions!!

```python
sigma_v = 4 # units : mm/frame_spacing, i.e. we only need to convert length units. Numerical value may be different for each age after unit conversion!
sigma_t = 1 # units : rds. Same for all ages
L = 50 # units : mm. Note that numerical value may differ for each age after unit conversion

# proximity parameters. These are different for each age as they scale with bodylength!
# need to convert these values from mm to grid units!
repulsion_radius = [4, 7, 10, 12] #mm
optimal_proximity = [6, 10.5, 15, 18] #mm
proximity_decay = 0.1 #(mm)^-1
```

```python
window_size = 5 #grid units
sampling_fraction = 0.6
```

### Circular grid constraint function

```python
def circular_constraint_func(grid, c_x, c_y, R):
    ind = ((grid["x"] - c_x) ** 2 + (grid["y"] - c_y) ** 2) < R**2
    return grid.loc[ind]
```

# Calculate and plot `proximity`, `vicsek` and `velocityDiffusion` predictors, and `nextStepExponential` score for each age

# 2wks

```python
length_conversion = grid_size_agewise[0]/arena_size_agewise[0] #grid_units / mm
grid_constraint_params = {
    "c_x": fish_2wks_object.grid_size / 2,
    "c_y": fish_2wks_object.grid_size / 2,
    "R": fish_2wks_object.grid_size / 2,
}

predictor_kwargs = {
    "proximity": {
        "interaction_length": L * length_conversion,
        "optimal_distance": optimal_proximity[0] * length_conversion,
        "repulsion_radius": repulsion_radius[0] * length_conversion,
        "proximity_decay": proximity_decay / length_conversion,
    },

    "vicsek": {
        "interaction_length" : L* length_conversion,
        "sigma_v" : sigma_v * length_conversion,
        "sigma_t" : sigma_t,
        "dt" : 1
    },

    "velocityDiffusion": {
        "sigma_v" : sigma_v * length_conversion,
        "sigma_t" : sigma_t,
        "dt" : 1
    }
}

local_windows_kwargs = {
    "window_size": window_size,
    "sampling_fraction": sampling_fraction,
    "skip_incomplete_frames": True,
    "grid_constraint" : circular_constraint_func,
    "grid_constraint_params" : grid_constraint_params

}

score_kwargs = {
    "nextStepExponential": {
            "decay_length" : 2
        }
}

fish_2wks_derivedDF = ft.derive_predictors_and_scores(
    fish_2wks_object,
    local_windows_kwargs,
    predictor_kwargs=predictor_kwargs,
    score_kwargs=score_kwargs,
    dropna=False,
    add_scaled_values=False)
```
